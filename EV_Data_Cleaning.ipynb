```python
# This shows every row that has at least one missing value anywhere in it
blank_rows = df[df.isnull().any(axis=1)]
blank_rows
```

```python
cols_to_fill = ['County', 'City', 'Postal Code', 'Electric Utility', '2020 GEOID']

for col in cols_to_fill:
    # Convert column to string first so 'Unknown' text fits without errors, then fill
    df[col] = df[col].astype(str).fillna('Unknown')
```

```python
df['Vehicle Location'] = df['Vehicle Location'].fillna('Unknown')
```

```python
median_range = df['Electric Range'].median()
df['Electric Range']=df['Electric Range'].fillna(median_range)
```

```python
df['Legislative District'] = df['Legislative District'].astype(str).fillna('Unknown')
```

```python
import pandas as pd

# List of columns that had errors
cols_with_errors = ['2020 GEOID', 'Legislative District', 'Postal Code']

for col in cols_with_errors:
    # 1. Turn the errors into NaN (missing values)
    # 2. Fill those missing values with a safe placeholder (e.g., 0 or 'Unknown')
    # 3. Overwrite the exact same column directly
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# Now, export normally without adding any extra files or columns
df.to_csv('cleaned_ev_data.csv', index=False)
```
